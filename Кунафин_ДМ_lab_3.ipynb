```python
import pandas as pd
from google.colab import files
import io

uploaded = files.upload()

for fn in uploaded.keys():
  print('User uploaded file "{name}" with length {length} bytes'.format(
      name=fn, length=len(uploaded[fn])))

  # Read the CSV file into a pandas DataFrame, specifying the tab delimiter
  df = pd.read_csv(io.StringIO(uploaded[fn].decode('utf-8')), sep='\t')
  display(df.head())
```

Check if the variable `df` exists and is a pandas DataFrame, then display the head and info of the dataframe.

## Очистка данных

### Subtask:
Обработайте пропущенные значения и некорректные данные.


**Reasoning**:
Examine the DataFrame for missing values, drop the column with only missing values, and clean the column names by removing leading/trailing spaces or special characters.



```python
# Examine missing values
print("Missing values per column:")
print(df.isnull().sum())

# Drop the 'Unnamed: 15' column
df = df.drop('Unnamed: 15', axis=1)
print("\nDataFrame after dropping 'Unnamed: 15' column:")
display(df.head())

# Clean column names
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace('\t', '')
print("\nCleaned column names:")
print(df.columns)
```

**Reasoning**:
Analyze the unique values and their frequencies in the `occupation` and `work_mode` columns to identify any inconsistencies or incorrect entries that might need cleaning or standardization.



```python
# Analyze unique values and their frequencies in 'occupation'
print("Unique values and frequencies in 'occupation' column:")
print(df['occupation'].value_counts())

# Analyze unique values and their frequencies in 'work_mode'
print("\nUnique values and frequencies in 'work_mode' column:")
print(df['work_mode'].value_counts())
```

## Преобразование типов данных

### Subtask:
Преобразуйте строковые значения в числовые, если это необходимо для алгоритмов машинного обучения.


**Reasoning**:
Identify and convert categorical columns to numerical using one-hot encoding.



```python
# Identify columns with object data types
object_columns = df.select_dtypes(include=['object']).columns
print(f"Object columns to convert: {list(object_columns)}")

# Apply one-hot encoding to categorical columns
df = pd.get_dummies(df, columns=object_columns, drop_first=True)

# Display the first few rows of the updated DataFrame
display(df.head())
```

## Выбор признаков

### Subtask:
Определите, какие столбцы будут использоваться для анализа или моделирования.


**Reasoning**:
Identify the numerical and one-hot encoded columns for analysis and modeling, excluding identifier columns.



```python
# Identify columns suitable for analysis and modeling.
# Exclude 'user_id' as it is an identifier.
# All other numerical and one-hot encoded columns are suitable.
selected_features = [col for col in df.columns if col != 'user_id']

# Create a new DataFrame with selected features
df_selected = df[selected_features].copy()

print(f"Number of selected features: {len(selected_features)}")
display(df_selected.head())
```

## Масштабирование данных

### Subtask:
Примените масштабирование к числовым признакам, если это требуется для выбранных алгоритмов.


**Reasoning**:
Apply standard scaling to the numerical features in the `df_selected` DataFrame as required for selected algorithms.



```python
from sklearn.preprocessing import StandardScaler

# Instantiate a StandardScaler object
scaler = StandardScaler()

# Fit the scaler to the df_selected DataFrame and transform it
df_scaled = scaler.fit_transform(df_selected)

# Convert the scaled NumPy array back to a DataFrame with original column names
df_scaled = pd.DataFrame(df_scaled, columns=df_selected.columns)

# Display the first few rows of the df_scaled DataFrame
display(df_scaled.head())
```

```python
from sklearn.model_selection import train_test_split

# Define features (X) for splitting. Since this is likely an unsupervised task or
# the target isn't specified yet, we split the entire scaled dataframe.
X = df_scaled

# Split the data into training and testing sets
X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)

# Print the shapes of the resulting training and testing sets
print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
```
